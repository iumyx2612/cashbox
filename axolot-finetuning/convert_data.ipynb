```python
import json 
import pandas as pd 


data_path = '/home/hoang.minh.an/anhalu-data/learning/cashbox/data/update_data/baseline_time_function_calling_v5.csv'
```

```python
df = pd.read_csv(data_path)
```

```python
df
```

```python
datasets = []
for i, row in df.iterrows():
    dataset = {
        'messages': 
            [
                {
                    'role': 'system',
                    'content': row['system']
                },
                {
                    'role': 'user',
                    'content': row['user']
                },
                {
                    'role': 'assistant',
                    'content': row['json']
                }
            ],
    }
    datasets.append(dataset) 
```

```python
with open('/home/hoang.minh.an/anhalu-data/learning/cashbox/data/update_data/baseline_time_function_calling_v5_axolot_format.json', 'w') as f:
    json.dump(datasets, f, indent=4, ensure_ascii=False)
```
